```python
print("------------------------------------------------------ YRC BANK --------------------------------------------------------------------------")

pin=int(input("Please make your pin:"))
balance=int(input("Enter balance amount:"))
print("thanks for opening account!")
ans=input("Do you want to withdraw cash?Yes or No?")
if ans=='Yes':
    check=int(input("Please enter your pin:"))
    if check==pin:
        withdraw=int(input("Enter amount to withdraw:"))
        if balance>=withdraw:
            print("sure , here is the cash!")
        else:
            print("Balance insuffiecient")
    else:
        print("Wrong pin, go home")
else:
    print("Thank you , bye!")
```
