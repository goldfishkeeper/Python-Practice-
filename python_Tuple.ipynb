```python
#To make a tuple
tuple1=tuple(('apple','banana','grapes'))
print(tuple1)
```

```python
#To make single item tuple
tuple2=('mango',)
print(tuple2)
print(type(tuple2))
```

```python
#To convert tuple to list 
list1=list(tuple1)
print(list1)
```

```python
#To add/remove items in tuple
print('The input tuple is ',tuple2)
list2=list(tuple2)
print(list2)
list2.append('guava')
list2.append('pineapple')
print(list2)
list2.remove('mango')
print(list2)
tuple3=tuple(list2)
print('Edited tuple is',tuple3)
```

```python
#To join tuples
tuple4=('watermelon','strawberry')
print('Before joining tuple was:',tuple4)
tuple4+=tuple3
print('After joining tuple is:',tuple4)
```

```python
#To delete a tuple
tuple5=(1,2,3,4,5)
print(tuple5)
del(tuple5)
print(tuple5)  #This gives error as tuple5 is deleted in previous line...
```
