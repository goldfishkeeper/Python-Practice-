```python
#To make a list 
list1=list(("apple","banana","guava"))
print(list1)
```

```python
#To change an item
list1[1]="watermelon"
print(list1)
```

```python
#To add an item
list1.append("mango")
print(list1)
```

```python
#To add an item at specific index
list1.insert(2,"pineapple")
print(list1)
```

```python
#To remove an item
list1.remove('apple')
print(list1)
```

```python
#To remove an item using index number
list1.pop(3)
print(list1)
```

```python
#To combine two lists 
list2=[1,2,3,4]
list1.extend(list2)
print(list1)
```

```python
#To print range of items
print(list1[2:5])
```

```python
#To sort list
list3=[1,3,4,6,7]
list3.sort()#ascending
print(list3)
list3.sort(reverse=True)
print(list3)
```

```python
#To clear the list
list3.clear()
print(list3)
```
