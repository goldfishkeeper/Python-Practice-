```python
import pandas as pd
```

```python
#To read raw csv
c=pd.read_csv("titanic.csv")
print(c)
```

```python
#To print info
print(c.info())
```

```python
#To find null values location
c.isnull()
```

```python
#To find columns with null values
c.isnull().any()
```

```python
#To find sum of null values
c.isnull().sum()
```

```python
#To drop null values
n=c.dropna()
print(n)
n.isnull().sum()
```

```python
#To fill empty cells
k=c.fillna("none")
print(k)
k.isnull().sum()
```
